**MATPLOTLIB**
* is a python library used for creating graphs and visualizing data
* pyplot: is a module used to create and customize graphs in python
* plt.plot: is used to create a line graph in python
* show: Display the graph

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
plt.plot(months,sales)
plt.show()
```

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
sales=[200,30000,2000,45000,550000]
plt.plot(months,sales)
plt.show()
```

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,55]
profit=[5,8,6,12,15]

plt.plot(months,sales,label="sales") # Where the label means name the label
plt.plot(months,profit,label="profit")
plt.legend() # it means display the lines or data ina graph
plt.show()
```

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
plt.plot(
    months,
    sales,
    marker="o",
    color="violet",
    linestyle="solid"
)
plt.xlabel("months")
plt.ylabel("sales")
plt.title("Monthly Sales")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
plt.bar(months,sales)

plt.xlabel("months")
plt.ylabel("sales")
plt.title("Monthly Sales")
plt.grid(True)
plt.bar(
    months,
    sales,
    color="#00FF00"
)
plt.show()
```

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
plt.scatter(months,sales) # is used to show the relationship b/w to numerical variables using points

plt.xlabel("months")
plt.ylabel("sales")
plt.title("Monthly Sales")
plt.grid(True)
plt.scatter(
    months,
    sales,
    color="#FF1493"
)
plt.show()
```

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
plt.hist(months)        # is used to create a histogram to show the distribution of numerical data

plt.xlabel("months")
plt.ylabel("sales")
plt.title("Monthly Sales")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
Marks=[45,34,56,78,89,12,34,90,98,76,43,23,55]
plt.hist(Marks)
plt.title("Marks Distribution")

plt.xlabel("Marks")
plt.ylabel("Fequent")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
categories=["Food","Travel","Shopping","Others"]
amount=[10000,7000,4000,2000]

plt.pie(amount,labels=categories)
plt.legend()
plt.title("Monthly Expenses")
plt.show()
```
