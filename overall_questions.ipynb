**Question 1 : online store data**
* An online store has recorded the number of units sold for different product.
 product=["Laptop","Phones","Tablets,"Watchs"]

 the manager wants to compare the performance of the products and easily identify which product sold the most.
 * Task: visualize the data in a suitable way

```python
import seaborn as sns
import matplotlib.pyplot as plt
product=["Laptop","Phones","Tablets","Watch"]
sales=[50,80,40,70]
plt.pie(sales,labels=product,autopct="%4.2f%%")
plt.title("Online Store Sales")
plt.show()
sns.barplot(y=sales,x=product)
```

**Question 2 : Website visitor**


A company recorded its website visitors during a working week.
days = ["Mon","Tue","Wed","Thu","Fri"]
visitors = [120,150,180,160,210]
The marketing team wants to understand how visitor activity changed throughout the week.
Task : Create a suitable visualization ton show this change


```python
days = ["Mon","Tue","Wed","Thu","Fri"]
visitors = [120,150,180,160,210]
plt.plot(days,visitors)
plt.xlabel("days")
plt.ylabel("visitors")
plt.title("Website Visitor")
plt.show()
```

**Question 3: Monthly Expenses**
* categories=["Food","Travel","Rent","Shopping"]

```python
categories=["Food","Travel","Rent","Shopping"]
expense=[90000,50000,4000,70000]
plt.pie(expense,labels=categories,autopct="%5.2f%%")
plt.title("Monthly Expenses")
plt.show()
```
