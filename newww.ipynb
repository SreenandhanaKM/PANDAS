```python
import pandas as pd
sales=pd.Series([2000,1200,300,400,700],
                index=["Monday","Tuesday","wednesday","Thrusday","Friday"])
print(sales)
print("maximum sales",sales.max())
print("minimum sales",sales.min())
print(sales[sales<1000])
```

```python
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())
print(ar.nunique())   #count of unique number
```

```python
print(ar.value_counts())
```

```python
data=[None,20,30,None,50,60,None,80]
ar=pd.Series(data)
print(ar)
print(ar.isnull()) # check whether a value is missing and returns true if its missing
print(ar.notnull()) # check whether a value is not missing and return true if it contains a value
```
