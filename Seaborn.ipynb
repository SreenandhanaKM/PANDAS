**Seaborn**
* used for data visualization
* it is built on top of matplotlib and make it easier to create attractive and informative statisitical graphs
* import seaborn as sns

```python
import seaborn as sns
import matplotlib.pyplot as plt
data={

"Product":["Laptop","Phone","Tablet","Laptop","Phone","Tablet"],
"Sales":[50,80,40,70,90,60],
"Price":[50000,30000,22000,55000,3200,22000],
"Months":["Jan","Feb","Mar","Jan","Feb","Mar"]}
sns.barplot(x="Product",y="Sales",data=data)
plt.show()

sns.countplot(x="Product",data=data)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data={

"Product":["Laptop","Phone","Tablet","Laptop","Phone","Tablet"],
"Sales":[50,80,40,70,90,60],
"Price":[50000,30000,22000,55000,3200,22000],
"Months":["Jan","Feb","Mar","Jan","Feb","Mar"]}
sns.barplot(x="Product",y="Sales",data=data) # show bar
plt.show()

sns.countplot(x="Product",data=data)
plt.show()


```

```python
sns.countplot(x="Product",data=data) # show the number of observations in each category
plt.show()
```

```python
sns.boxplot(x="Product",y="Sales",data=data) # shows the distribution,median,and spread of numerical data
plt.show()
```

```python
sns.scatterplot(x="Product",y="Sales",data=data) # show the relationship b/w two numerical variables using points
plt.show()
```

```python
sns.lineplot(x="Product",y="Sales",data=data) # show trends or changes in data using connected lines
plt.show()
```
