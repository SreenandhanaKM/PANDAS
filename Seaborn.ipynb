**Seaborn**
* used for data visualization
* it is built on top of matplotlib and make it easier to create attractive and informative statisitical graphs
* import seaborn as sns

```python
import seaborn as sns
import matplotlib.pyplot as plt
data={

"Product":["Laptop","Phone","Tablet","Laptop","Phone","Tablet"],
"Sales":[50,80,40,70,90,60],
"Price":[50000,30000,22000,55000,3200,22000],
"Months":["Jan","Feb","Mar","Jan","Feb","Mar"]}
sns.barplot(x="Product",y="Sales",data=data)
plt.show()

sns.countplot(x="Product",data=data)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data={

"Product":["Laptop","Phone","Tablet","Laptop","Phone","Tablet"],
"Sales":[50,80,40,70,90,60],
"Price":[50000,30000,22000,55000,3200,22000],
"Months":["Jan","Feb","Mar","Jan","Feb","Mar"]}
sns.barplot(x="Product",y="Sales",data=data) # show bar
plt.show()

sns.countplot(x="Product",data=data)
plt.show()


```

```python
sns.countplot(x="Product",data=data) # show the number of observations in each category
plt.show()
```

```python
sns.boxplot(x="Product",y="Sales",data=data) # shows the distribution,median,and spread of numerical data
plt.show()
```

```python
sns.scatterplot(x="Product",y="Sales",data=data) # show the relationship b/w two numerical variables using points
plt.show()
```

```python
sns.lineplot(x="Product",y="Sales",data=data) # show trends or changes in data using connected lines
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = np.random.randn(1000)
plt.figure(figsize=(8,6))
sns.histplot(data,bins=40,kde=True,color="green",edgecolor="blue")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Distribution Of Random Data Using Seaborn")
plt.grid(axis="y",alpha=0.75)
plt.show()
```

```python
product=["Laptop","Phone","Tablet"]
sales=[50,80,40]

plt.pie(sales,labels=product,autopct="%4.2f%%")
plt.title("Sales Distribution")
plt.show()
```

```python
months=["Jan","Feb","Mar","Apr","May"]
sales=[10,50,30,80,50]
plt.plot(months,sales)
plt.fill_between(months,sales) # fills the area between two lines in a graph with color
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sales=[
    [10,20,30,40],
    [15,25,35,30],
    [20,30,40,35]
]
sns.heatmap(sales,annot=True,cmap="Blues") # show data values using different colors to represent their intensity or magnitude
plt.show()
```
